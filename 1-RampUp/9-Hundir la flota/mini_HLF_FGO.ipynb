## Mini Hundir la Flota
#### Primeros pasos
Vamos a empezar a programar trozos de código que nos servirán para el HLF:

* Crear un Tablero como una lista de listas de 10x10 que contenga caracteres espacio: " "
* Insertar dos barcos sustituyendo en el Tablero los espacios en blanco por el caracter "B":
    * Uno de 4 casillas en horizontal: casillas [1][0], [1][1], [1][2], [1][3]
    * Uno de 3 casillas en vertical:   casillas [3][3], [4][3], [5][3]
* Escribir una celda de código separada de las anteriores para simular que el usuario está efectuando un disparo: 
    *  Solicite al usuario dos coordenadas `i,j` mediante dos inputs
    *  Compruebe si esas coordenadas están ocupada por una casilla barco (que haya una en "B")
        * Si hay una "B":
            * Cambiar el contenido de esa posición por una "x"
            * Imprimir por pantalla el texto "Tocado en posición `i,j`"
        * Si hay un " ":
            * Cambiar el contenido de esa posición por una "o"
            * Imprimir por pantalla el texto "Agua"
    * Imprimir por pantalla el Tablero
* Para comprobar que funciona bien la celda de disparar ejecutarla varias veces metiendo distintas coordenadas

### Bonus
* En la celda de disparar, si el usuario acierta a tocar un barco, vuelve a disparar
* Resuélvelo con funciones:
    * Una para solicitar una coordenada válida a la que llamarás dos veces, una par i y otra par j
    * Una `disparar` que devuelva `True` si el usuario ha acertado o `False` si no.
    


```python
# Para crear un tablero vamos a hacer listas de 10x10, con espacios vacíos representados por un espacio en blanco " ":
def crear_tablero():
    tablero = [["~" for _ in range(10)] for _ in range(10)]
    return tablero

# Imprimir tablero:
def imprimir_tablero(tablero):
    for fila in tablero:
        fila_visible = []
        for celda in fila:
            if celda == "B":
                fila_visible.append("~")  # oculta barcos con olas
            else:
                fila_visible.append(celda)
        print(" ".join(fila_visible))
        
# Crear el tablero:
tablero = crear_tablero()
# Insertar dos barcos sustituyendo el espacio en blanco por "B":
# Barco horizontal de 4 casillas (fila 1)
tablero[1][0] = "B"
tablero[1][1] = "B"
tablero[1][2] = "B"
tablero[1][3] = "B"

# Barco vertical de 3 casillas (columna 3)
tablero[3][3] = "B"
tablero[4][3] = "B"
tablero[5][3] = "B"
imprimir_tablero(tablero)

def pedir_coordenada(mensaje):
    while True:
        try:
            coord = int(input(mensaje))  # intenta convertir a número
            if 0 <= coord <= 9:
                return coord
            else:
                print("Entre 0 y 9")
        except ValueError:               # si no es número, avisa
            print("Por favor, introduce un número")
## Función para disparar, que pide coordenadas al usuario y marca el tablero con "x" si acierta o "o" si falla:

def disparar(tablero):
    i = pedir_coordenada("Fila: ")
    j = pedir_coordenada("Columna: ")
    if tablero[i][j] == "B":
        tablero[i][j] = "x"
        print(f"Tocado en posición {i},{j}")
        return True
    else:
        tablero[i][j] = "o"
        print("Agua")
        return False

# Juego en si, con un bucle que se repite mientras el jugador acierte, y que muestra el tablero al finalizar:
while disparar(tablero):
    print("¡Tocado! Vuelves a disparar 🎯")
    imprimir_tablero(tablero)

for fila in tablero:
    print(" ".join(fila))
    
```
