## Ejercicio 3
Crea una lista nueva con todas las combinaciones de las siguientes dos listas:
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

```['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']```

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]
print(ejer_3_1,ejer_3_2)
resultado_ejer_3 = []
for palabra_1 in ejer_3_1:
    for palabra_2 in ejer_3_2:
        resultado_ejer_3.append(palabra_1 + " " + palabra_2)
print(resultado_ejer_3)
````

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
print(ejer_5)
ejer_5_copy = ejer_5.copy()
for x in ejer_5:
    if x == 3:
        ejer_5_copy.remove(x)

print(ejer_5, ejer_5_copy)
````

## Ejercicio 10
1. Convierte la tupla del ejercicio 9 en una lista
2. Convierte la tupla del ejercicio 9 en un set
3. Convierte la tupla del ejercicio 9 en un diccionario. Usa también los índices

````python
ejer_9 = (3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3, 5, 2, 4, 7, 9, 4, 2, 4, 3, 3, 4, 6, 7)
print(type(ejer_9))
# 1. Convierte la tupla anterior en una lista
lista_9_1 = list(ejer_9)
print(lista_9_1, type(lista_9_1))
# 2. Convierte la tupla anterior en un set
set_9_1 = set(ejer_9)
print(set_9_1, type(set_9_1))
# 4. Convierte la tupla anterior en un diccionario. Usa también los índices
dicc_9_1 = {}

for i, value in enumerate(ejer_9):
    dicc_9_1[i] = value

print(dicc_9_1, type(dicc_9_1))
````

## Ejercicio 11
Convierte la siguiente lista en un diccionario
> ```ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]
# Directamente sin pensar:
dicc_11 = dict(ejer_11)
print(dicc_11, type(dicc_11)) 
## Aquí vemos que se ha convertido en diccionario, pero ha sobreescrito los valores de las claves que se llaman igual, por eso sólo tenemos 3 claves con 3 valores solo.
````

````python
dicc_11_2 = {}

for clave, valor in ejer_11:
    print(clave, valor)
    if clave not in dicc_11_2:
        dicc_11_2[clave] = []
    dicc_11_2[clave].append(valor)
        
print(dicc_11_2, type(dicc_11_2))
## Con este bucle for conseguimos que dentro de cada clave tenga los valores correspondientes.
````

## Ejercicio 12
1. Crea una lista ordenada ascendente con las claves del diccionario
2. Crea otra lista ordenada descendente con los valores
3. Añade una nueva clave/valor
4. Busca la clave igual a 2 dentro del diccionario
5. Itera la clave y el valor del diccionario con un único for
> ```ejer_12 = {4:78, 2:98, 8:234, 5:29}```

````python
ejer_12 = {4:78, 2:98, 8:234, 5:29}
# 1. Crear una lista ordenada ascendente con las claves del diccionario:
lista_11_1 = list(ejer_12.keys())
print(lista_11_1, type(lista_11_1))
lista_11_1 = sorted(list(ejer_12.keys()))
print(lista_11_1)
# 2. Crear una lista ordenada descendente con los valores:
lista_11_2 = list(ejer_12.values())
lista_11_2 = sorted(list(ejer_12.values()), reverse=True)
print(lista_11_2)
# 3. Añadir una nueva clave/valor:
ejer_12[7] = 22
print(ejer_12)
# 4. Buscar la clave igual a 2 dentro del diccionario:
print(ejer_12[2])
# 5. Itera la clave y el valor del diccionario con un único for:
for clave, valor in ejer_12.items(): 
    print(clave, valor)
````
