### Se pide, usando regresión lineal:

Dibujar con una línea la relación que hay entre la altura y la edad de los alumnos de esta clase: 

```python
lista_alumnos = [("Leonardo S", 24, 1.82), 
                 ("Piero T", 25, 1.71), 
                 ("Marta B", 35, 1.66), 
                 ("Silvia P", 37, 1.63), 
                 ("Faro Z", 29, 1.90), 
                 ("Miguel N", 27, 1.80), 
                 ("Alejandro M", 28, 1.70), 
                 ("Cristina M", 32, 1.60), 
                 ("Francisco P", 36, 1.74), 
                 ("Jorge D", 45, 1.72), 
                 ("Jesús L", 41, 1.65), 
                 ("Marta G", 30, 1.65), 
                 ("Jennifer S", 40, 1.60), 
                 ("Diego I", 39, 1.80), 
                 ("Antonio C", 23, 1.77), 
                 ("Juan M", 32, 1.75), 
                 ("David S", 27, 1.70), 
                 ("Antonio J", 34, 1.80), 
                 ("Carlos H", 27, 1.77), 
                 ("Erik U", 28, 1.70), 
                 ("Marcos L", 35, 1.80)] 
```

```python
# Transformamos la lista de alumnos en un DataFrame de pandas
df = pd.DataFrame(lista_alumnos, columns=["Nombre", "Edad", "Altura"])
df.head(10)
```

```python
# Análisis descriptivo
df.describe().T
```

```python
sns.regplot(data=df, x="Edad", y="Altura", 
            scatter_kws={"color": "steelblue"}, 
            line_kws={"color": "red"})
plt.title("Edad vs Altura con regresión")
plt.show()
```

### Entrena un modelo de regresión lineal

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
X = df[['Edad']]
X.values
```

```python
y = df[['Altura']]
y.values
```

```python
# Entrenar el modelo con el 80% de los datos y testearlo con el 20% restante
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 42)
```

```python
print("Total features shape:", X.shape)
print("Train features shape:", X_train.shape)
print("Train target shape:", y_train.shape)
print("Test features shape:", X_test.shape)
print("Test target shape:", y_test.shape)
```

```python
#Importar Linear Regression Algorithm
from sklearn.linear_model import LinearRegression
```

```python
lm = LinearRegression()
lm.fit(X_train, y_train)
```

```python
plt.scatter(df["Edad"], df["Altura"])
plt.plot(df["Edad"], lm.predict(df[["Edad"]]), color="red", label="Regresión lineal")
plt.xlabel("Edad"); plt.ylabel("Altura"); plt.title("Regresión Lineal: Edad vs Altura")
plt.legend(); plt.show()
```

### ¿Qué error se comete? Calcula los errores que está cometiendo tu modelo de forma manual con python. Aparte, usa las métricas de sklearn MAE, MAPE, MSE y el RMSE.

```python
predictions = lm.predict(X_test)
predictions
```

```python
# Errores que comete el modelo forma manual:
MAE  = np.mean(np.abs(y_test - predictions))
MAPE = np.mean(np.abs((y_test - predictions) / y_test)) * 100
MSE  = np.mean((y_test - predictions) ** 2)
RMSE = np.sqrt(MSE)
print('MAE:', MAE)
print('MAPE:', MAPE)
print('MSE:', MSE)
print('RMSE:', RMSE)
```

```python
from sklearn import metrics
print('MAE:', metrics.mean_absolute_error(y_test, predictions))
print('MSE:', metrics.mean_squared_error(y_test, predictions))
print('RMSE:', np.sqrt(metrics.mean_squared_error(y_test, predictions)))
```
