2. Sabemos que podemos pintar gráficas de dos formas: la figura incluye los ejes o teniendo figura y ejes por separado.

Usando solo una figura, usa numpy para los valores del eje X entre 0 y 5. 

Pinta dos gráficas en dos cajas distintas, a la izquierda una recta con pendiente positiva de 3 que pase por (0,0) y a la derecha una recta con pendiente negativa de 3 que pase por (0,-5). 

Elige la precisión en el eje X que desees.  

```python
x = np.linspace(0, 5, 100)

y1 = 3*x              # pendiente positiva
y2 = -3*x - 5         # pendiente negativa

plt.figure(figsize=(10,4))

plt.subplot(1,2,1) # subplot hace que se puedan poner varios gráficos en la misma figura, en este caso 2 gráficos, uno al lado del otro (1 fila, 2 columnas, gráfico 1)
plt.plot(x, y1)

plt.subplot(1,2,2) # gráfico 2, en la misma figura, pero en la segunda posición (1 fila, 2 columnas, gráfico 2)
plt.plot(x, y2);
```

3. Fija el eje X entre 0 y 5 y el eje Y entre -15 y 15

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(x, y1)
plt.xlim(0,5) # fijar eje x entre 0 y 5
plt.ylim(-15,15) # fijar eje y entre -15 y 15

plt.subplot(1,2,2)
plt.plot(x, y2)
plt.xlim(0,5)
plt.ylim(-15,15);
```

4. Llama al eje X "eje X", al eje Y "eje Y" y pon de títulos "recta sube" y "recta baja". Muestra dos etiquetas de ejeX pero solo una de eje Y.

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(x, y1)
plt.title("recta sube") # titulo del gráfico
plt.xlabel("eje X") # etiqueta del eje x
plt.ylabel("eje Y") # etiqueta del eje y

plt.subplot(1,2,2)
plt.plot(x, y2)
plt.title("recta baja")
plt.xlabel("eje X");
```

13. Vamos a introducir pandas. Crea un DataFrame con las columnas "Year" de valores 2015, 2016, 2017, 2018, 2019 y la columna "Sold_items_A" de valores 1000, 3500, 4000, 5500, 7000

```python
df = pd.DataFrame({
    'Year': [2015, 2016, 2017, 2018, 2019],
    'Sold_items_A': [2000, 3000, 4000, 5500, 7000]
})
df # Comprobar que se ha creado bien el DataFrame
```

14. Ahora pintamos un gráfico de línea con las ventas respecto al año. Pon un título y etiquetas en los ejes.
Cambia el estilo a 'seaborn-white'. La línea debe ser con rayas y verde.

```python
#plt.style.use('seaborn-white')  # versión antigua de seaborn-white
plt.style.use('seaborn-v0_8-white') # cambio de estilo a seaborn-white

plt.plot(
    df['Year'],
    df['Sold_items_A'],
    linestyle='--', # discontinua
    color='green' # verde
)

plt.title("Ventas por año")
plt.xlabel("Año")
plt.ylabel("Ventas");
```

15. Otro departamento B ha vendido en esos años 2000, 3100, 5000, 4000, 6000 unidades. Incluye esa columna en el DataFrame y pinta en la misma gráfica las dos líneas. B es una línea punteada y roja. Muestra una leyenda abajo a la derecha.

```python
df['Sold_items_B'] = [2000, 3100, 5000, 4000, 6000]
df # Comprobar que se ha introducido bien la segunda columna.

```

```python
plt.plot(df['Year'], df['Sold_items_A'],
         linestyle='-', color='blue', label='Dept A') # línea azul y contínua

plt.plot(df['Year'], df['Sold_items_B'],
         linestyle='--', color='red', label='Dept B') # línea roja y discontinua

plt.legend(loc='lower right') # leyenda abajo a la derecha
plt.title("Ventas por año")
plt.xlabel("Año")
plt.ylabel("Ventas");
```

19. Vamos a pintar un histograma de unas alturas generadas aleatoriamente. Crea 300 alturas entre 150 y 190 cm.

Pista: https://pandas.pydata.org/pandas-docs/version/0.23.4/generated/pandas.DataFrame.plot.html

```python
alturas = np.random.randint(150, 190, 300) # se han generado 300 alturas aleatorias entre 150 y 190 cm con np.random.randint.

df_alturas = pd.DataFrame({'Altura': alturas}) # se ha creado un DataFrame con una columna llamada 'Altura' que contiene las alturas generadas aleatoriamente.

df_alturas.plot(kind='hist', bins=20, edgecolor='black') # gráfico de histograma con 20 bins y borde
plt.xlabel("Altura (cm)") 
plt.ylabel("Frecuencia")
plt.title("Histograma de Alturas");
```
