# 23. Mapas de calor
### Calcula las correlaciones del primer DataFrame original df
### Pista: las correlaciones se calculan con el método corr()
### Recuerda: las correlaciones miden entre dos variables, si varío una, cuánto me varía la otra (cómo están de relacionadas)

```python
# Calcular la matriz de correlación con el dataframe origina, sólo se puede correlacionar con variables numéricas, así que pongo numeric_only = True.
corr = pokemon.corr(numeric_only=True)
corr
```

```python
# Hacer el heatmap para representar y visualizar mejor la matriz de correlación, con los valores y un mapa de calor.
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Heatmapo de la Matriz de Correlación de los Pokemons");
```

### ¿la correlación entre qué valores se encuentra?
### la autocorrelación es la correlación de una variable consigo misma, ¿qué aprecias?
### Encuentra dos variables que veas que están fuertemente relacionadas según la correlación

La correlación varía entre -1 y 1, siendo la correlación -1 una correlación perfecta y negativa, y 1 una correlación perfecta y positiva. Si la correlación es cero, es que es nula y no hay correlación entre las variables.,
Por lo tanto, una variable consigo misma tiene una correlación perfecta siempre.
La relación más alta que se observa es entre la defensa especial y el total (que es la suma de todas las variables de ataque, defensa, ataque especial, defensa especial, etc) con una correlación de 0.77, y es una correlación positiva y fuerte.
La otra relación con una correlación positiva y alta es el total con el ataque especial (correlación 0.68) y el total con el ataque (correlación 9.67).
Por lo tanto, tiene todo el sentido decir que estás variables están relacionadas unas con otras (no son independientes) y por tanto tienen una correlación tan alta.
