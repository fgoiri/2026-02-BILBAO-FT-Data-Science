# Extrae los dominios de los correos con una expresión regular

```python
import re

emails = '''
miguellc@thebridgeschool.tech
miguel.cruz@uc3m.es
miguel-12-2000@mi-correo.net
miguellc@s@thebridgeschool.es
'''
```

Ejemplo:
* @thebridgeschool.tech
* @uc3m.es
* @mi-correo.net
* @thebridgeschool.es

```python
pattern = r'(@[\w.-]+\.[a-zA-Z]{2,})$'

dominios = re.findall(pattern, emails, re.MULTILINE)
print(dominios)
```
