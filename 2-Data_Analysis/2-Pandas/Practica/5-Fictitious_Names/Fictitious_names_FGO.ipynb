```python
# Copio esta clase por si luego quiero que me queden más bonitos los prints
class display(object):
    """Representador HTML de múltiples objetos"""
    template = """<div style="float: left; padding: 10px;">
    <p style='font-family:"Courier New", Courier, monospace'>{0}</p>{1}
    </div>"""
    def __init__(self, *args):
        self.args = args
        
    def _repr_html_(self):
        return '\n'.join(self.template.format(a, eval(a)._repr_html_())
                         for a in self.args)
    
    def __repr__(self):
        return '\n\n'.join(a + '\n' + repr(eval(a))
                           for a in self.args)
    
```

### Step 2. Create the 3 DataFrames based on the following raw data

```python
raw_data_1 = {
        'subject_id': ['1', '2', '3', '4', '5'],
        'first_name': ['Alex', 'Amy', 'Allen', 'Alice', 'Ayoung'], 
        'last_name': ['Anderson', 'Ackerman', 'Ali', 'Aoni', 'Atiches']}

raw_data_2 = {
        'subject_id': ['4', '5', '6', '7', '8'],
        'first_name': ['Billy', 'Brian', 'Bran', 'Bryce', 'Betty'], 
        'last_name': ['Bonder', 'Black', 'Balwner', 'Brice', 'Btisan']}

raw_data_3 = {
        'subject_id': ['1', '2', '3', '4', '5', '7', '8', '9', '10', '11'],
        'test_id': [51, 15, 15, 61, 16, 14, 15, 1, 61, 16]}
```

### Step 3. Assign each to a variable called data1, data2, data3

```python
# Hago el ejercicio 2 directamente con el ejercicio 3, hago los data frames con el raw data anterior, y lo guardo directamente en las variables
data1 = pd.DataFrame(raw_data_1)
data1.info() # Quiero ver que se me han importado bien los datos y los tipos de datos.
```

```python
data2 = pd.DataFrame(raw_data_2)
data2.info() # Quiero ver que se me han importado bien los datos y los tipos de datos.
```

### Step 4. Join the two first dataframes along rows and assign to all_data

```python
# Unir los dos primeros dataframes por medio de filas 
all_data = pd.concat([data1, data2], axis = 0).reset_index(drop=True) # con reset_index: drop = True, le digo que resetee el índice.
all_data
# Tambien se puede usar ignore_index
```

### Step 7. Merge all_data and data3 along the subject_id value

```python
data_all_data3 = pd.merge(all_data, data3, on = "subject_id") # Unión de los dos dataframes por medio de la columna subject_id.
data_all_data3
```

### Step 8. Merge only the data that has the same 'subject_id' on both data1 and data2

```python
data_1y2 = pd.merge(data1, data2, on = "subject_id", how='inner') # no hace falta poner el how, pero para saber que hago un inner.
data_1y2 # Con el inner sólo coge las coincidencias, con el outer, todo de ambos y rellena los vacíos con NaNs.
```
